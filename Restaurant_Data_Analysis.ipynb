### 9. Outlier Analysis


*   The graph showed many values outside the normal range




```python
df.boxplot(column='Average Cost for two')
plt.show()
```

```python
df['Average Cost for two'].quantile([0.25, 0.50, 0.75])
```

```python
Q1 = df['Average Cost for two'].quantile(0.25)
Q3 = df['Average Cost for two'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower :", lower)
print("Upper :", upper)
```

#### So values above 1,375 were identified as statistical outliers under this overall IQR calculation

```python
outliers = df[
    (df['Average Cost for two'] < lower) |
    (df['Average Cost for two'] > upper)]
outliers[['Restaurant Name','Country Code','City',
          'Average Cost for two','Currency'  ]].head(20)
```

### Important


#### *   Because this dataset contains multiple currencies, I didn't blindly delete these values.

#### *   I considered the currency column before interpreting them.
